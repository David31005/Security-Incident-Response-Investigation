```python

import pandas as pd

url = "https://raw.githubusercontent.com/David31005/Security-Incident-Response-Investigation/main/security_events.csv"

logs = pd.read_csv(url)

print("SECURITY INCIDENT INVESTIGATION")
print("-------------------------------")
print(logs.to_string(index=False))
```

```python
print("SOC SECURITY ALERT ANALYSIS")
print("---------------------------")

failed = logs[(logs["Event_Type"] == "Login") & (logs["Status"] == "Failed")]
counts = failed.groupby("Source_IP").size()

for ip, count in counts.items():
    if count >= 4:
        print(f"ALERT: Possible brute-force activity from {ip}")
        print(f"Failed login attempts: {count}")

        successful = logs[
            (logs["Source_IP"] == ip) &
            (logs["Event_Type"] == "Login") &
            (logs["Status"] == "Success")
        ]

        changes = logs[
            (logs["Source_IP"] == ip) &
            (logs["Event_Type"] == "Privilege_Change")
        ]

        if not successful.empty:
            print("WARNING: Successful login recorded from the same IP")

        if not changes.empty:
            print("CRITICAL INDICATOR: Privilege change recorded from the same IP")

        print("Recommended action: Investigate account activity and verify authorization.")
```

```python

print("SECURITY INCIDENT TIMELINE")
print("--------------------------")

# Convert timestamps to datetime values
logs["Timestamp"] = pd.to_datetime(logs["Timestamp"])

# Focus on the suspicious source IP
suspicious_ip = "203.0.113.45"

incident = logs[
    logs["Source_IP"] == suspicious_ip
].sort_values("Timestamp")

# Display chronological security events
for _, event in incident.iterrows():
    print(
        f"{event['Timestamp']} | "
        f"{event['Username']} | "
        f"{event['Event_Type']} | "
        f"{event['Status']} | "
        f"{event['Destination']}"
    )

print("--------------------------")
print("Total suspicious-source events:", len(incident))
```
